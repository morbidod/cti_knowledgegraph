Step 2:\
Costruzione del grafo\
Modelliamo le entità Company, BusinessSector, NISSector, AtecoCode e ClassificationEvidence collegate da predicati semantici:

```python
EX = Namespace("https://data.company-nis2.it/resource/")
ONT = Namespace("https://data.company-nis2.it/ontology/")

def build_knowledge_graph(df: pd.DataFrame) -> Graph:
    g = Graph()
    g.bind("ex", EX)
    g.bind("ont", ONT)
    g.bind("rdfs", RDFS)

    for _, row in df.iterrows():
        guid = str(row["guid"]).strip()
        denominazione = str(row["denominazione"]).strip()
        comp_uri = EX[f"company_{guid}"]
        ev_uri = EX[f"evidence_{guid}"]
        
        # 1. Nodo Azienda / Entità Soggetto
        g.add((comp_uri, RDF.type, ONT.Company))
        g.add((comp_uri, ONT.guid, Literal(guid, datatype=XSD.string)))
        g.add((comp_uri, ONT.denominazione, Literal(str(row["denominazione"]), datatype=XSD.string)))
        g.add((comp_uri, ONT.needsHumanReview, Literal(bool(row["necessita_revisione_umana"]), datatype=XSD.boolean)))
        
        if row["qualifica_nis"] != "N/D":
            g.add((comp_uri, ONT.legalQualification, Literal(str(row["qualifica_nis"]), datatype=XSD.string)))

        # 2. Nodo Evidenza di Classificazione (Decision Evidence)
        g.add((ev_uri, RDF.type, ONT.ClassificationEvidence))
        g.add((ev_uri, ONT.agentType, Literal(str(row["agent_type"]), datatype=XSD.string)))
        g.add((ev_uri, ONT.decisionSource, Literal(str(row["fonte_decisionale"]), datatype=XSD.string)))
        g.add((ev_uri, ONT.motivation, Literal(str(row["motivazione"]), datatype=XSD.string)))
        g.add((comp_uri, ONT.evaluatedBy, ev_uri))

        # 3. Nodo Codice ATECO
        ateco = str(row["ateco"]).strip()
        if ateco and ateco != "N/D":
            ateco_clean = ateco.replace(" ", "_").replace(".", "_")
            a_uri = EX[f"ateco_{ateco_clean}"]
            g.add((a_uri, RDF.type, ONT.AtecoCode))
            g.add((a_uri, ONT.code, Literal(ateco, datatype=XSD.string)))
            g.add((a_uri, ONT.descrizioneAteco, Literal(row["descrizione_ateco"], datatype=XSD.string)))
            g.add((a_uri, ONT.hasDiscrepancy, Literal(bool(row["incoerenza_ateco"]), datatype=XSD.boolean)))
            g.add((comp_uri, ONT.hasAssignedAteco, a_uri))

        # 4. Nodo Settore Target di Business
        settore = str(row["settore_target"]).strip()
        if settore and settore != "N/D":
            s_clean = settore.replace(" ", "_").replace("'", "")
            s_uri = EX[f"sector_{s_clean}"]
            g.add((s_uri, RDF.type, ONT.BusinessSector))
            g.add((s_uri, RDFS.label, Literal(settore, datatype=XSD.string)))
            g.add((ev_uri, ONT.suggestsBusinessSector, s_uri))

        # 5. Nodo Settore NIS 2
        nis_sec = str(row["settore_nis2"]).strip()
        if nis_sec and nis_sec != "Fuori Perimetro NIS 2":
            nis_clean = nis_sec.replace(" ", "_")[:35]
            nis_uri = EX[f"nis_{nis_clean}"]
            g.add((nis_uri, RDF.type, ONT.NISSector))
            g.add((nis_uri, RDFS.label, Literal(nis_sec, datatype=XSD.string)))
            g.add((comp_uri, ONT.hasNISSector, nis_uri))

    return g

kg_graph = build_knowledge_graph(df_classificato)
print(f"Grafo costruito con successo: {len(kg_graph)} triple RDF.")
```

Step 3: Estrazione SPARQL & definizione funzione di Visualizzazione Grafica Puntuale\
Estraiamo i nodi e gli archi via SPARQL per una specifica entità (identificata dal suo guid) e generiamo la vista interattiva con PyVis

```python
import html
import os
from pyvis.network import Network
from IPython.display import HTML

def get_entity_subgraph_sparql(graph: Graph, guid: str) -> dict:
    guid_clean = str(guid).strip()
    company_uri = f"https://data.company-nis2.it/resource/company_{guid_clean}"
    
    query = f"""
    PREFIX ont: <https://data.company-nis2.it/ontology/>
    PREFIX rdfs: <http://www.w3.org/2000/01/rdf-schema#>

    SELECT ?denominazione ?needsReview ?source ?motivation ?ateco ?descrizioneAteco ?sector ?nisSector
    WHERE {{
        BIND(<{company_uri}> AS ?company)
        
        # Recupero con il predicato effettivamente popolato nel grafo
        OPTIONAL {{ ?company ont:denominazione ?denominazione . }}
        
        OPTIONAL {{ ?company ont:needsHumanReview ?needsReview . }}
        
        OPTIONAL {{
            ?company ont:evaluatedBy ?ev .
            OPTIONAL {{ ?ev ont:decisionSource ?source . }}
            OPTIONAL {{ ?ev ont:motivation ?motivation . }}
            OPTIONAL {{ 
                ?ev ont:suggestsBusinessSector ?s . 
                ?s rdfs:label ?sector . 
            }}
        }}

        OPTIONAL {{ 
            ?company ont:hasAssignedAteco ?a . 
            ?a ont:code ?ateco . 
           OPTIONAL {{ ?a ont:descrizioneAteco ?descrizioneAteco . }}
        }}
        
        OPTIONAL {{ 
            ?company ont:hasNISSector ?nis . 
            ?nis rdfs:label ?nisSector . 
        }}
    }}
    """
    res = list(graph.query(query))
    if not res:
        return None
        
    r = res[0]
    return {
        "guid": guid_clean,
        "denominazione": str(r.denominazione) if r.denominazione is not None else "N/D",
        "needs_review": bool(r.needsReview) if r.needsReview is not None else False,
        "source": str(r.source) if r.source is not None else "N/D",
        "motivation": str(r.motivation) if r.motivation is not None else "Nessuna motivazione registrata",
        "ateco": str(r.ateco) if r.ateco is not None else "N/D",
        "descrizioneAteco": str(r.descrizioneAteco) if r.descrizioneAteco is not None else "N/D",
        "sector": str(r.sector) if r.sector is not None else "N/D",
        "nis_sector": str(r.nisSector) if r.nisSector is not None else "Fuori Perimetro",
    }



def render_agent_result_graph(graph: Graph, guid: str, filename: str = "temp_graph.html"):
    data = get_entity_subgraph_sparql(graph, guid)
    if not data:
        print(f"Nessun dato trovato per il GUID: {guid}")
        return None

    # Usiamo cdn_resources='remote' senza notebook=True per generare un HTML standard e autosufficiente
    net = Network(height="480px", width="100%", directed=True, bgcolor="#ffffff", cdn_resources="remote")

    # 1. Nodo Centrale Soggetto
    c_color = "#dc3545" if data["needs_review"] else "#1d4ed8"
    c_id = f"comp_{guid[:8]}"
    net.add_node(
        c_id,
        label=f"Soggetto\n{guid[:8]}...",
        title=f"GUID: {guid}\nRevisione Umana: {data['needs_review']}",
        color=c_color,
        size=26,
        shape="dot"
    )

    # 2. Nodo Evidenza Decisionale
    ev_id = f"ev_{guid[:8]}"
    fonte_breve = (data['source'][:25] + '...') if len(data['source']) > 25 else data['source']
    net.add_node(
        ev_id,
        label=f"Fonte:\n{fonte_breve}",
        title=f"Motivazione:\n{data['motivation']}",
        color="#f59e0b",
        size=20,
        shape="box"
    )
    net.add_edge(c_id, ev_id, label="ont:evaluatedBy")

    # 3. Settore Target
    if data["sector"] != "N/D":
        s_id = f"sec_{data['sector'][:10]}"
        net.add_node(s_id, label=f"Settore:\n{data['sector']}", color="#10b981", shape="ellipse", size=22)
        net.add_edge(ev_id, s_id, label="ont:suggestsSector")

    # 4. Codice ATECO
    if data["ateco"] != "N/D":
        net.add_node(data["ateco"], label=f"ATECO: {data['ateco']}-{data['descrizione_ateco']}", color="#64748b", shape="box")
        net.add_edge(c_id, data["ateco"], label="ont:hasAssignedAteco")

    # 5. Settore NIS 2
    if data["nis_sector"] != "Fuori Perimetro":
        nis_id = f"nis_{data['nis_sector'][:10]}"
        net.add_node(nis_id, label=f"NIS 2:\n{data['nis_sector']}", color="#0dcaf0", shape="box", size=20)
        net.add_edge(c_id, nis_id, label="ont:hasNISSector")

    net.set_options('{"physics": {"forceAtlas2Based": {"gravitationalConstant": -70}}, "edges": {"smooth": false}}')

    # Salviamo l'HTML generato da PyVis
    net.save_graph(filename)

    # Leggiamo il codice HTML e facciamo l'escape dentro srcdoc
    with open(filename, "r", encoding="utf-8") as f:
        raw_html = f.read()

    escaped_html = html.escape(raw_html)
    
    iframe_code = f"""
    <iframe
        style="width: 100%; height: 520px; border: 1px solid #d1d5db; border-radius: 8px;"
        srcdoc="{escaped_html}"
        frameborder="0">
    </iframe>
    """
    return HTML(iframe_code)
```

Funzione di salvataggio del grafo per visualizzazione via browser

```python
import os
import webbrowser
from pyvis.network import Network

def export_and_open_graph(graph: Graph, guid: str, filename: str = "grafo_soggetto.html"):
    data = get_entity_subgraph_sparql(graph, guid)
    if not data:
        print(f"Nessun dato trovato per il GUID: {guid}")
        return None

    # Creazione della rete senza flag notebook
    net = Network(height="750px", width="100%", directed=True, bgcolor="#ffffff", cdn_resources="remote")

    # 1. Nodo Centrale Soggetto (Rosso se richiede review, Blu altrimenti)
    c_color = "#dc3545" if data["needs_review"] else "#1d4ed8"
    c_id = f"comp_{guid[:8]}"
    net.add_node(
        c_id,
        label=f"Soggetto\n{data['denominazione']}",
        title=f"GUID: {guid}\nRevisione Umana: {data['needs_review']}",
        color=c_color,
        size=28,
        shape="dot"
    )

    # 2. Nodo Evidenza Decisionale
    ev_id = f"ev_{guid[:8]}"
    fonte_breve = (data['source'][:25] + '...') if len(data['source']) > 25 else data['source']
    net.add_node(
        ev_id,
        label=f"Fonte:\n{fonte_breve}",
        title=f"Motivazione:\n{data['motivation']}",
        color="#f59e0b",
        size=22,
        shape="box"
    )
    net.add_edge(c_id, ev_id, label="ont:evaluatedBy")

    # 3. Settore Target
    if data["sector"] != "N/D":
        s_id = f"sec_{data['sector'][:10]}"
        net.add_node(s_id, label=f"Settore:\n{data['sector']}", color="#10b981", shape="ellipse", size=24)
        net.add_edge(ev_id, s_id, label="ont:suggestsSector")

    # 4. Codice ATECO
    if data["ateco"] != "N/D":
        net.add_node(data["ateco"], label=f"ATECO: {data['ateco']}{data['descrizioneAteco']}-", color="#64748b", shape="box")
        net.add_edge(c_id, data["ateco"], label="ont:hasAssignedAteco",            title=f"📌 MOTIVAZIONE DELLA SCELTA:\n{data['motivation']}",
            width=2,
            hoverWidth=4)

    # 5. Settore NIS 2
    if data["nis_sector"] != "Fuori Perimetro":
        nis_id = f"nis_{data['nis_sector'][:10]}"
        net.add_node(nis_id, label=f"NIS 2:\n{data['nis_sector']}", color="#0dcaf0", shape="box", size=22)
        net.add_edge(c_id, nis_id, label="ont:hasNISSector")

    net.set_options('{"physics": {"forceAtlas2Based": {"gravitationalConstant": -70}}, "edges": {"smooth": false}}')

    # Salvataggio su file HTML locale con percorso assoluto
    abs_path = os.path.abspath(filename)
    net.save_graph(abs_path)
    print(f"Grafo salvato con successo in: {abs_path}")

    # Apertura automatica nel browser
    webbrowser.open(f"file://{abs_path}")
    return abs_path
```
